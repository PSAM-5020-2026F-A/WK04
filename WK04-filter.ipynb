```python
from PIL import Image as PImage
from image_utils import get_pixels, make_image
```

### Define Helping Function

```python
def color_distance(c0, c1):
  return ((c0[0] - c1[0])**2 + (c0[1] - c1[1])**2 + (c0[2] - c1[2])**2) ** 0.5
```

### Filter by Color

```python
img = PImage.open("./data/hog.jpg")
display(img)
```

```python
REFERENCE_COLOR = [200, 200, 0]
THRESHOLD = 80

filtered_pixels = []
for r,g,b in get_pixels(img):
  if color_distance([r, g, b], REFERENCE_COLOR) < THRESHOLD:
    filtered_pixels.append([r,g,b])
  else:
    l = sum([r,g,b]) // 3
    filtered_pixels.append([l,l,l])

display(make_image(filtered_pixels, img.width))
```
